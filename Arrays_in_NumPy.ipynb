```python
# An ndarray (N-dimensional array) is the core, powerful data structure provided by the NumPy library
# 2D Array (Matrix / Grid)
import numpy as np
import time
arr = np.array([[1,2,3],[4,5,6]])
print(arr.shape)  # Output: (2, 3) -> 2 rows, 3 columns
print(arr.ndim)   # Output: 2 (Number of dimensions)
print(arr.dtype)  # Output: int64 (Data type of the items)
print(arr.size)   # Output: 6 (Total number of elements)
```

```python
import numpy as np
import time

# Create 1 million items in a regular Python list and a NumPy array
size = 1000000
python_list = list(range(size))
numpy_array = np.arange(size)

# Time how long it takes a Python list to add 2 to every item
start_time = time.time()
python_list = [x + 2 for x in python_list]
print(f"Python List took: {time.time() - start_time:.5f} seconds")

# Time how long it takes NumPy to do the exact same thing
start_time = time.time()
numpy_array = numpy_array + 2  # This is called Array Broadcasting!
print(f"NumPy Array took: {time.time() - start_time:.5f} seconds")
```
