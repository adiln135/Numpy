```python
import numpy as np
```

```python
arr = np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])
```

```python
print(arr[0,1])      # 20 (row 0, column 1)
print(arr[2,2])      # 90 (row 2, column 2)
print(arr[-1,-3])    # 70 (last row, 2nd last column)
print(arr[-3,-3])    # 10 (first row, first column)
```

```python
print(arr[0:2, 1:3 ])
 # 0:2 (First position): Selects the rows. It starts at index 0 and stops before index 2, giving you row 0 and row 1.
 # 1:3 (Second position): Selects the columns. It starts at index 1 and stops before index 3, giving you column 1 and column 2.
```

```python
print(arr[::2, ::2])  # (row 0, row 2, row 4, etc.). (column 0, column 2, column 4, etc.).
```

```python
slice_arr = arr[0, :2]
slice_arr[0] = 999
print(arr[0])
```
